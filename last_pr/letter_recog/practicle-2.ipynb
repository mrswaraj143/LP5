````python
"""This practical is about **Multiclass Classification using Deep Neural Networks (DNN)** using the **OCR Letter Recognition Dataset**.

The goal is:

* Input → features of a letter image
* Output → predict which letter it is (`A` to `Z`)

Since there are **26 possible classes**, this is called:

# Multiclass Classification

---

# 1. Importing Libraries

```python
import numpy as np
import pandas as pd
```

## What these do

### NumPy (`np`)

Used for:

* mathematical operations
* arrays
* handling predictions

### Pandas (`pd`)

Used for:

* reading dataset
* handling tables/dataframes

---

# 2. Reading Dataset

```python
df = pd.read_csv("letter-recognition.data")
```

## What happens here?

The dataset file is loaded into a dataframe named `df`.

Each row represents:

* one alphabet letter
* along with 16 numerical features describing it

---

# 3. Viewing Dataset

```python
df.head()
```

Shows first 5 rows.

```python
df.tail()
```

Shows last 5 rows.

```python
df.shape
```

Shows:

```python
(rows, columns)
```

For this dataset:

* 20,000 rows
* 17 columns

Why 17?

* 1 column → output letter
* 16 columns → input features

---

# 4. Splitting Input and Output

```python
x = df.iloc[:,1:]
y = df.iloc[:,0]
```

This is VERY IMPORTANT.

---

# Understanding `iloc`

`iloc` means:

# Integer Location

Used to select rows/columns using index numbers.

---

# Dataset Structure

| Column 0 | Column 1 | Column 2 | ... |
| -------- | -------- | -------- | --- |
| Letter   | Feature1 | Feature2 | ... |

---

# `x = df.iloc[:,1:]`

Meaning:

```python
[row selection , column selection]
```

### `:` before comma

means:

* select ALL rows

### `1:`

means:

* select columns from index 1 till end

So:

```python
x = input features
```

Contains:

* 16 numerical columns

---

# `y = df.iloc[:,0]`

Means:

* all rows
* only column 0

This is:

```python
A, B, C, D ...
```

Target/output labels.

---

# 5. Encoding Labels

```python
from sklearn.preprocessing import LabelEncoder
```

Neural networks cannot understand letters like:

```python
A B C
```

They only understand numbers.

---

## Creating Encoder

```python
encoder = LabelEncoder()
```

---

## Transforming Labels

```python
y = encoder.fit_transform(y)
```

Converts:

| Letter | Number |
| ------ | ------ |
| A      | 0      |
| B      | 1      |
| C      | 2      |
| ...    | ...    |
| Z      | 25     |

Now output becomes numeric.

---

# 6. Train-Test Split

```python
from sklearn.model_selection import train_test_split
```

---

## Why split data?

We divide dataset into:

| Part          | Purpose           |
| ------------- | ----------------- |
| Training Data | Learn patterns    |
| Testing Data  | Check performance |

---

## Splitting

```python
x_train,x_test,y_train,y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

---

# Understanding Parameters

## `test_size=0.2`

20% data for testing.

80% for training.

If total rows = 20,000:

| Train  | Test  |
| ------ | ----- |
| 16,000 | 4,000 |

---

# `random_state=42`

Controls random splitting.

Without this:

* every run gives different train/test data

With this:

* same split every time

Useful for:

* reproducibility
* debugging
* comparing models

---

# Important Doubt

You asked earlier:

> "Doesn't random_state cause learning of training/testing data?"

NO.

It only fixes the random selection process.

The model STILL:

* trains only on training data
* never sees test labels during training

So no cheating happens.

---

# 7. Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
```

---

# Why scaling?

Features may have different ranges.

Example:

| Feature | Values  |
| ------- | ------- |
| height  | 1-10    |
| area    | 100-500 |

Large values dominate learning.

So we normalize features.

---

# StandardScaler Formula

It transforms data using:

genui{"math_block_widget_always_prefetch_v2":{"content":"z = \frac{x-\mu}{\sigma}"}}

Where:

* (x) = original value
* (\mu) = mean
* (\sigma) = standard deviation

Result:

* mean becomes 0
* standard deviation becomes 1

---

# Scaling Training Data

```python
scalar = StandardScaler()

x_train = scalar.fit_transform(x_train)
```

---

# `fit_transform()`

Two steps:

1. Learn mean/std
2. Transform data

---

# VERY IMPORTANT MISTAKE IN YOUR CODE

You wrote:

```python
x_test = scalar.fit_transform(x_test)
```

This is WRONG.

Why?

Because test data should NOT calculate its own mean/std.

Correct:

```python
x_test = scalar.transform(x_test)
```

---

# Correct Version

```python
x_train = scalar.fit_transform(x_train)
x_test = scalar.transform(x_test)
```

---

# Why?

We want:

* training statistics used everywhere

Otherwise:

* data leakage happens
* evaluation becomes unrealistic

---

# 8. Importing TensorFlow

```python
import tensorflow as tf
```

TensorFlow is a deep learning framework.

Used to create:

* neural networks
* train models
* perform predictions

---

# 9. Creating Neural Network

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

---

# Sequential Model

```python
model = Sequential()
```

Sequential means:

* layers connected one after another

---

# 10. First Hidden Layer

```python
model.add(Dense(128,activation='relu',input_shape=(16,)))
```

---

# Understanding This Layer

## Dense

Fully connected layer.

Every neuron connected to all previous neurons.

---

# `128`

Number of neurons.

This layer learns patterns from inputs.

---

# `input_shape=(16,)`

Because dataset has:

* 16 input features

So:

* 16 input neurons

---

# ReLU Activation

f(x)=\max(0,x)

If:

* x > 0 → output x
* x < 0 → output 0

Why used?

* avoids vanishing gradient
* faster training
* introduces non-linearity

---

# 11. Second Hidden Layer

```python
model.add(Dense(64,activation='relu'))
```

Another hidden layer:

* 64 neurons
* learns deeper patterns

---

# 12. Output Layer

```python
model.add(Dense(26,activation='softmax'))
```

---

# Why 26 neurons?

Because:

* 26 letters (A-Z)

Each neuron corresponds to one class.

---

# Softmax Activation

Converts outputs into probabilities.

Example:

| Letter | Probability |
| ------ | ----------- |
| A      | 0.01        |
| B      | 0.90        |
| C      | 0.03        |

All probabilities sum to 1.

---

# Softmax Formula

\sigma(z_i)=\frac{e^{z_i}}{\sum_{j=1}^{K}e^{z_j}}

---

# 13. Compiling Model

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

---

# Optimizer = Adam

Adam updates weights efficiently.

It combines:

* Momentum
* RMSProp

Very popular optimizer.

---

# Loss Function

```python
sparse_categorical_crossentropy
```

Used because:

* multiclass classification
* labels are integers (0-25)

---

# Accuracy Metric

Measures:

* correct predictions percentage

---

# 14. Training Model

```python
history = model.fit(
    x_train,
    y_train,
    epochs=20,
    batch_size=32,
    validation_split=0.1
)
```

---

# Epochs

```python
epochs=20
```

One epoch:

* complete pass through training data

20 epochs:

* dataset processed 20 times

---

# Batch Size

```python
batch_size=32
```

Training happens in small groups of 32 samples.

Benefits:

* faster
* memory efficient

---

# Validation Split

```python
validation_split=0.1
```

10% of training data used for validation.

Validation checks:

* how model performs during training

---

# 15. Evaluating Model

```python
loss,accuracy = model.evaluate(x_test,y_test)
```

Checks performance on unseen test data.

---

# Printing Accuracy

```python
print("Accuracy: ",accuracy)
```

Usually gets:

* around 95%+

---

# 16. Making Predictions

```python
prediction = model.predict(x_test)
```

For every sample:

* outputs probabilities for 26 letters

---

# Example

```python
print(prediction[0])
```

Might show:

```python
[0.01 0.00 0.95 0.02 ...]
```

---

# Finding Highest Probability

```python
predicted_class = np.argmax(prediction[0])
```

`argmax()` returns:

* index of largest value

Example:

```python
[0.01 0.00 0.95]
```

Largest value index = 2

So predicted class = 2

---

# Converting Number Back to Letter

```python
letter = encoder.inverse_transform([predicted_class])
```

Converts:

```python
2 → C
```

---

# Final Flow of Practical

```text
Dataset
   ↓
Split input/output
   ↓
Encode labels
   ↓
Train-test split
   ↓
Feature scaling
   ↓
Build neural network
   ↓
Train model
   ↓
Evaluate accuracy
   ↓
Predict letters
```

---

# Why DNN Works Well Here

OCR letter recognition is complex because:

* different handwriting styles
* shapes vary
* many patterns

Deep Neural Networks:

* learn hidden patterns automatically
* perform well on image-like structured data

---

# Important Viva Questions

## What is multiclass classification?

Classification with more than two classes.

---

## Why use softmax?

To convert outputs into probabilities.

---

## Why use LabelEncoder?

Neural networks need numeric labels.

---

## Why scaling is important?

Prevents large-value features from dominating.

---

## Difference between ReLU and Softmax?

| ReLU              | Softmax                |
| ----------------- | ---------------------- |
| Hidden layers     | Output layer           |
| Removes negatives | Produces probabilities |

---

## Why sparse categorical crossentropy?

Because labels are integer encoded.
"""
````
