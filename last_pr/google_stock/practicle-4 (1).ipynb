````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
````

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dropout
from tensorflow.keras.layers import LSTM
````

````python
from sklearn.preprocessing import MinMaxScaler
````

````python
df = pd.read_csv('GOOGL.csv')
df
````

````python
df.head()
````

````python
df.tail()
````

````python
df.shape
````

````python
training_Set = df.iloc[:,1:2].values
````

````python
scaler = MinMaxScaler(feature_range=(0,1))
training_set_scaled = scaler.fit_transform(training_Set)
````

````python
x_train = []
y_train = []

for i in range(60,len(training_set_scaled)):
    x_train.append(training_set_scaled[i-60:i,0])
    y_train.append(training_set_scaled[i,0])
````

````python
x_train = np.array(x_train)
y_train = np.array(y_train)
````

````python
x_train = np.reshape(
    x_train,
    (x_train.shape[0],x_train.shape[1],1)
)
````

````python
model = Sequential()
````

````python
model.add(
    LSTM(
        units=50,
        return_sequences=True,
        input_shape = (x_train.shape[1],1)
    )
)
````

````python
model.add(Dropout(0.2))
````

````python
model.add(
    LSTM(
        units=50,
        return_sequences=False
    )
)
````

````python
model.add(Dropout(0.2))
````

````python
model.add(Dense(units=1))
````

````python
model.compile(
    optimizer='adam',
    loss='mean_squared_error'
)
````

````python
model.fit(
    x_train,
    y_train,
    epochs=20,
    batch_size=32
)
````

````python
predicted_stock_pprice = model.predict(x_train)
````

````python
predicted_stock_price = scaler.inverse_transform(predicted_stock_pprice)
````

````python
predicted_stock_pprice
````

````python
plt.plot(
    scaler.inverse_transform(y_train.reshape(-1,1)),
    color='red',
    label='real google stock price'
)

plt.plot(
    predicted_stock_price,
    color='blue',
    label='predicted stock price'
)

plt.title('google stock price prediction')

plt.xlabel('time')

plt.ylabel('google stock price')

plt.legend()

plt.show()
````

````python
"""This practical is about using a **Recurrent Neural Network (RNN)**, specifically an **LSTM**, to predict Google stock prices using past stock data.

The idea is:

* We give the model the **previous 60 days stock prices**
* The model learns patterns
* Then it predicts the **next day's stock price**

---

# 1. Importing Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

These are basic libraries:

| Library    | Purpose                      |
| ---------- | ---------------------------- |
| pandas     | Reading and handling dataset |
| numpy      | Mathematical operations      |
| matplotlib | Plotting graphs              |

---

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dropout
from tensorflow.keras.layers import LSTM
```

These are deep learning components.

| Component  | Purpose                               |
| ---------- | ------------------------------------- |
| Sequential | Creates neural network layer-by-layer |
| Dense      | Fully connected output layer          |
| Dropout    | Prevents overfitting                  |
| LSTM       | Special RNN layer for time series     |

---

```python
from sklearn.preprocessing import MinMaxScaler
```

Used for **normalization/scaling**.

---

# 2. Loading Dataset

```python
df = pd.read_csv('GOOGL.csv')
```

Reads Google stock dataset.

---

```python
df
df.head()
df.tail()
df.shape
```

These are inspection commands.

| Command | Meaning              |
| ------- | -------------------- |
| df      | Shows full dataframe |
| head()  | First 5 rows         |
| tail()  | Last 5 rows          |
| shape   | Rows and columns     |

---

# 3. Selecting Data

```python
training_Set = df.iloc[:,1:2].values
```

## Important

`iloc[:,1:2]`

Means:

* `:` → all rows
* `1:2` → column index 1 only

Suppose dataset:

| Date | Open | High | Low |
| ---- | ---- | ---- | --- |
| 1    | 100  | 110  | 90  |

Then:

```python
df.iloc[:,1:2]
```

selects:

| Open |
| ---- |
| 100  |

---

## Why `1:2` and not `1`?

Because:

```python
1:2
```

returns a **2D array**

Needed by scaler and neural network.

---

```python
.values
```

Converts dataframe into numpy array.

---

# 4. Feature Scaling

```python
scaler = MinMaxScaler(feature_range=(0,1))
```

Scales values between 0 and 1.

---

## Why scaling is needed?

Stock prices may be:

```python
1000
1200
1500
```

Neural networks work better with smaller values.

So scaler converts:

| Original | Scaled |
| -------- | ------ |
| 1000     | 0.2    |
| 1500     | 0.8    |

---

```python
training_set_scaled = scaler.fit_transform(training_Set)
```

Two things happen:

| Function  | Meaning            |
| --------- | ------------------ |
| fit       | Learns min and max |
| transform | Scales values      |

---

# 5. Creating Time Series Data

```python
x_train = []
y_train = []
```

These store:

| Variable | Meaning          |
| -------- | ---------------- |
| x_train  | Previous 60 days |
| y_train  | Next day price   |

---

# 6. Creating Sequences

```python
for i in range(60,len(training_set_scaled)):
```

Loop starts from 60 because:

We need previous 60 days to predict next day.

---

# Example

Suppose data:

```python
1 2 3 4 5 6
```

If using previous 3 values:

| Input | Output |
| ----- | ------ |
| 1 2 3 | 4      |
| 2 3 4 | 5      |
| 3 4 5 | 6      |

Same logic here using 60 days.

---

```python
x_train.append(training_set_scaled[i-60:i,0])
```

This stores previous 60 stock prices.

---

## Example

If `i=60`

```python
training_set_scaled[0:60]
```

stores day 0 to day 59.

---

```python
y_train.append(training_set_scaled[i,0])
```

Stores actual next day price.

So:

| Previous 60 days | Predict |
| ---------------- | ------- |
| Day 1 → Day 60   | Day 61  |

---

# 7. Convert Lists to Arrays

```python
x_train = np.array(x_train)
y_train = np.array(y_train)
```

Neural networks require numpy arrays.

---

# 8. Reshaping for LSTM

```python
x_train = np.reshape(
    x_train,
    (x_train.shape[0],x_train.shape[1],1)
)
```

This is VERY IMPORTANT.

---

# Why reshape?

LSTM expects 3D input:

```python
(samples, timesteps, features)
```

---

# Here:

| Dimension | Meaning                     |
| --------- | --------------------------- |
| samples   | Number of training examples |
| timesteps | 60 previous days            |
| features  | 1 feature (stock price)     |

---

# Example Shape

Suppose:

```python
(100,60)
```

means:

* 100 examples
* each has 60 values

After reshape:

```python
(100,60,1)
```

Now LSTM understands:

* 100 samples
* 60 timesteps
* 1 feature

---

# 9. Creating Model

```python
model = Sequential()
```

Creates empty neural network.

---

# 10. First LSTM Layer

```python
model.add(
    LSTM(
        units=50,
        return_sequences=True,
        input_shape = (x_train.shape[1],1)
    )
)
```

---

# Understanding Parameters

## units=50

Number of neurons/memory cells.

More units = more learning capacity.

---

## input_shape

```python
(60,1)
```

Means:

* 60 timesteps
* 1 feature

---

## return_sequences=True

This means:

Return entire sequence to next LSTM layer.

Needed because another LSTM layer comes after this.

---

# LSTM Working

LSTM remembers previous information.

Unlike normal neural networks:

* RNN/LSTM has memory
* Good for sequential/time-series data

---

# 11. Dropout Layer

```python
model.add(Dropout(0.2))
```

Drops 20% neurons randomly during training.

---

# Why?

Prevents overfitting.

Without dropout:

Model may memorize training data.

---

# 12. Second LSTM Layer

```python
model.add(
    LSTM(
        units=50,
        return_sequences=False
    )
)
```

Second LSTM learns higher-level patterns.

---

## return_sequences=False

Now we only need final output.

So sequence is not returned.

---

# 13. Another Dropout

```python
model.add(Dropout(0.2))
```

Again prevents overfitting.

---

# 14. Output Layer

```python
model.add(Dense(units=1))
```

Final prediction layer.

Outputs:

* one stock price

---

# 15. Compiling Model

```python
model.compile(
    optimizer='adam',
    loss='mean_squared_error'
)
```

---

# Optimizer

```python
adam
```

Updates weights efficiently.

Most commonly used optimizer.

---

# Loss Function

```python
mean_squared_error
```

Used for regression problems.

---

## Formula

MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2

Measures difference between:

* actual price
* predicted price

Lower MSE = better predictions.

---

# 16. Training Model

```python
model.fit(
    x_train,
    y_train,
    epochs=20,
    batch_size=32
)
```

---

# epochs=20

Entire dataset passes through network 20 times.

---

# batch_size=32

32 samples processed together at once.

---

# During Training

Model:

1. predicts output
2. calculates error
3. updates weights
4. improves predictions

---

# 17. Prediction

```python
predicted_stock_pprice = model.predict(x_train)
```

Model predicts stock prices.

---

# 18. Inverse Scaling

```python
predicted_stock_price = scaler.inverse_transform(predicted_stock_pprice)
```

Remember:

Data was scaled between 0 and 1.

This converts predictions back to original stock prices.

---

# 19. Plotting Real Prices

```python
plt.plot(
    scaler.inverse_transform(y_train.reshape(-1,1)),
    color='red',
    label='real google stock price'
)
```

Plots actual prices in red.

---

# Why reshape?

Scaler expects 2D array.

---

# 20. Plotting Predicted Prices

```python
plt.plot(
    predicted_stock_price,
    color='blue',
    label='predicted stock price'
)
```

Plots predicted prices.

---

# 21. Graph Labels

```python
plt.title('google stock price prediction')
```

Graph title.

---

```python
plt.xlabel('time')
```

X-axis = time.

---

```python
plt.ylabel('google stock price')
```

Y-axis = stock price.

---

```python
plt.legend()
```

Shows labels.

---

```python
plt.show()
```

Displays graph.

---

# Overall Flow of Practical

```text
Dataset
   ↓
Select Stock Prices
   ↓
Scale Data
   ↓
Create 60-Day Sequences
   ↓
Build LSTM Model
   ↓
Train Model
   ↓
Predict Prices
   ↓
Compare Real vs Predicted
```

---

# Why LSTM is Used Instead of Normal Neural Network?

Because stock prices are sequential.

Today's price depends on previous days.

LSTM can remember past information.

---

# What Makes LSTM Special?

LSTM solves:

* short-term memory problem
* vanishing gradient problem in RNN

It can remember long-term dependencies.

---

# Viva Questions You May Get

## What is RNN?

Neural network designed for sequential data.

---

## What is LSTM?

Special type of RNN with memory cells.

---

## Why use scaling?

Improves training speed and accuracy.

---

## Why reshape into 3D?

LSTM expects:

```python
(samples,timesteps,features)
```

---

## Why use Dropout?

Prevents overfitting.

---

## Why use previous 60 days?

To learn trends and patterns.

---

## Why MSE loss?

Because stock prediction is a regression problem.

---

# Important Observation

Your code predicts on training data itself:

```python
model.predict(x_train)
```

In real projects:

* separate testing data is used
* otherwise results may look overly accurate

This practical mainly demonstrates:

* Time series preprocessing
* RNN/LSTM architecture
* Sequential prediction system
"""
````
