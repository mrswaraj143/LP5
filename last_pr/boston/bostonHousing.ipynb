````python
scalar = StandardScaler()
x_train = scalar.fit_transform(x_train)
x_test = scalar.transform(x_test)
````

````python
from sklearn.metrics import mean_squared_error,r2_score

mse = mean_squared_error(y_test, y_pred)
r2square = r2_score(y_test,y_pred)

print("MSE:", mse)
print("R2 Square: ",r2square)
````

````python
"""Boston Housing Price Prediction using Deep Neural Network (DNN)

This practical demonstrates how to perform **Linear Regression using a Deep Neural Network (DNN)** on the **Boston Housing Dataset** to predict house prices.

The goal is:

* Input → House features like crime rate, number of rooms, tax, etc.
* Output → Predict house price (`medv`)

---

# 1. Importing Libraries

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

import tensorflow as tf
```

## Explanation

### NumPy

```python
import numpy as np
```

Used for numerical operations and arrays.

---

### Pandas

```python
import pandas as pd
```

Used for handling datasets in tabular form.

---

### StandardScaler

```python
from sklearn.preprocessing import StandardScaler
```

Used to normalize data so that all features are on a similar scale.

---

### train_test_split

```python
from sklearn.model_selection import train_test_split
```

Used to divide dataset into:

* Training data
* Testing data

---

### TensorFlow

```python
import tensorflow as tf
```

TensorFlow is a deep learning framework used to build and train neural networks.

---

# 2. Loading Dataset

```python
df = pd.read_csv("BostonHousing.csv")
```

## Explanation

* Reads the Boston Housing dataset from CSV file.
* Stores dataset in dataframe `df`.

---

# 3. Display Dataset

```python
df
```

Dataset contains 14 columns.

## Features (Input Variables)

| Column  | Meaning                            |
| ------- | ---------------------------------- |
| crim    | Crime rate                         |
| zn      | Residential land proportion        |
| indus   | Industrial area proportion         |
| chas    | Charles River dummy variable       |
| nox     | Nitric oxide concentration         |
| rm      | Average rooms per house            |
| age     | Old house proportion               |
| dis     | Distance from employment centers   |
| rad     | Highway accessibility              |
| tax     | Property tax                       |
| ptratio | Pupil-teacher ratio                |
| b       | Black population proportion        |
| lstat   | Lower status population percentage |

---

## Target Column

| Column | Meaning            |
| ------ | ------------------ |
| medv   | Median house price |

---

# 4. Separating Input and Output

```python
x = df.iloc[:,:-1]
y = df.iloc[:,-1]
```

## Explanation

### Input Features

```python
x = df.iloc[:,:-1]
```

Selects all columns except last column.

So `x` contains:

* 13 independent variables

---

### Target Variable

```python
y = df.iloc[:,-1]
```

Selects last column (`medv`).

This is the value we want to predict.

---

# 5. Checking Dataset

```python
df.head()
```

Shows first 5 rows.

---

```python
df.tail()
```

Shows last 5 rows.

---

```python
df.shape
```

Output:

```python
(506, 14)
```

Meaning:

* 506 rows
* 14 columns

---

# 6. Splitting Dataset

```python
x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size=0.2,random_state=40
)
```

## Explanation

Dataset is divided into:

| Data     | Percentage |
| -------- | ---------- |
| Training | 80%        |
| Testing  | 20%        |

---

## Why Split Data?

### Training Data

Used to train the neural network.

### Testing Data

Used to check model accuracy on unseen data.

---

## Parameters

### `test_size=0.2`

20% data used for testing.

---

### `random_state=40`

Ensures same random split every time.

---

# 7. Feature Scaling

```python
scalar = StandardScaler()
x_train = scalar.fit_transform(x_train)
x_test = scalar.transform(x_test)
```

## Why Scaling is Needed?

Features have different ranges:

Example:

* `crim` may be small
* `tax` may be very large

Neural networks work better when values are normalized.

---

## StandardScaler Formula

genui{"math_block_widget_always_prefetch_v2":{"content":"z = \frac{x-\mu}{\sigma}"}}

Where:

* (x) = original value
* (\mu) = mean
* (\sigma) = standard deviation

---

## fit_transform()

```python
scalar.fit_transform(x_train)
```

* Calculates mean and standard deviation
* Transforms training data

---

## transform()

```python
scalar.transform(x_test)
```

Uses same scaling values on testing data.

---

# 8. Importing Deep Learning Model

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

## Explanation

### Sequential

Used to create neural network layer-by-layer.

---

### Dense

Represents fully connected neural network layers.

---

# 9. Creating Neural Network

```python
model = Sequential()
```

Creates empty neural network.

---

# 10. Adding Layers

## First Hidden Layer

```python
model.add(Dense(64,activation='relu',input_shape=(13,)))
```

## Explanation

### 64 Neurons

This layer contains 64 neurons.

---

### ReLU Activation

f(x)=\max(0,x)

If value is negative → output 0
If positive → same value

ReLU helps network learn non-linear relationships.

---

### `input_shape=(13,)`

Input layer accepts 13 features.

---

# Second Hidden Layer

```python
model.add(Dense(32,activation='relu'))
```

* 32 neurons
* ReLU activation

This layer extracts deeper patterns.

---

# Output Layer

```python
model.add(Dense(1))
```

## Explanation

Only 1 neuron because:

* We predict one value (house price)

No activation because regression outputs continuous values.

---

# 11. Warning Message

```python
UserWarning: Do not pass an input_shape...
```

This is not an error.

Keras recommends newer syntax:

```python
from tensorflow.keras import Input

model = Sequential([
    Input(shape=(13,)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1)
])
```

Your code still works correctly.

---

# 12. Compiling Model

```python
model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)
```

## Explanation

Before training, model must be compiled.

---

## Optimizer = Adam

```python
optimizer='adam'
```

Adam adjusts weights efficiently to reduce error.

---

## Loss Function = MSE

```python
loss='mse'
```

Used for regression problems.

Formula:

MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2

Where:

* (y_i) = actual value
* (\hat{y}_i) = predicted value

Lower MSE means better predictions.

---

## Metric = MAE

```python
metrics=['mae']
```

MAE = Mean Absolute Error

MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y}_i|

Shows average prediction error.

---

# 13. Training the Model

```python
model.fit(
    x_train,
    y_train,
    epochs=100,
    batch_size=16
)
```

## Explanation

This trains neural network.

---

## Parameters

### `epochs=100`

Entire dataset passes through model 100 times.

---

### `batch_size=16`

16 samples processed at one time.

---

# Understanding Training Output

Example:

```python
Epoch 1/100
loss: 543.5069 - mae: 21.9122
```

Initially:

* Error is very high

As epochs increase:

* Loss decreases
* MAE decreases

Meaning:

* Model learns patterns successfully.

---

# Final Training Result

```python
loss: 5.3901 - mae: 1.7055
```

Meaning:

* Very low prediction error
* Good training performance

---

# 14. Predicting House Prices

```python
y_pred = model.predict(x_test)
```

## Explanation

Uses trained model to predict house prices for testing data.

---

# 15. Evaluating Model

```python
from sklearn.metrics import mean_squared_error,r2_score
```

Used to measure performance.

---

# Mean Squared Error

```python
mse = mean_squared_error(y_test, y_pred)
```

Output:

```python
MSE: 10.17659604183867
```

Lower value means better prediction.

---

# R² Score

```python
r2square = r2_score(y_test,y_pred)
```

Output:

```python
R2 Square: 0.9102806241328694
```

## Formula

R^2 = 1 - \frac{\sum (y_i-\hat{y}_i)^2}{\sum (y_i-\bar{y})^2}

---

## Interpretation

| R² Value | Meaning       |
| -------- | ------------- |
| 0        | Poor model    |
| 1        | Perfect model |

Your model:

```python
0.91
```

Meaning:

* Model explains 91% variance
* Excellent prediction accuracy

---

# Final Output

```python
print("MSE:", mse)
print("R2 Square: ",r2square)
```

Output:

```python
MSE: 10.17
R2 Square: 0.91
```

---

# Complete Flow of Practical

```text
Load Dataset
      ↓
Separate Input & Output
      ↓
Split Dataset
      ↓
Scale Features
      ↓
Create DNN Model
      ↓
Add Hidden Layers
      ↓
Compile Model
      ↓
Train Model
      ↓
Predict Prices
      ↓
Evaluate Accuracy
```

---

# Why This is Called Linear Regression using DNN?

Traditional Linear Regression uses:

genui{"math_block_widget_always_prefetch_v2":{"content":"y = wx + b"}}

But here:

* Neural network learns weights automatically
* Hidden layers capture complex relationships
* Final output is still continuous numeric value

So this becomes:

* Regression problem solved using Deep Neural Network

---

# Viva Questions with Answers

## 1. Why do we use StandardScaler?

To normalize features so neural network trains efficiently.

---

## 2. Why ReLU activation?

ReLU solves vanishing gradient problem and speeds learning.

---

## 3. Why no activation in output layer?

Regression problems require continuous output values.

---

## 4. What is epoch?

One complete pass of training data through network.

---

## 5. What is batch size?

Number of samples processed together before weight update.

---

## 6. Why use MSE?

MSE penalizes larger errors more strongly.

---

## 7. What does R² score indicate?

How well predictions match actual values.

---

# Conclusion

In this practical:

* Boston Housing dataset was used
* Data preprocessing and scaling were performed
* A Deep Neural Network was created using TensorFlow/Keras
* Model was trained for regression
* House prices were predicted successfully
* Achieved:

  * Low MSE
  * High R² score (0.91)

This indicates the DNN model performs very well for house price prediction.
"""
````
